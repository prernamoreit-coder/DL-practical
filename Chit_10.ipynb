```python
# 10. Implement a CNN to extract features from grayscale hand gesture images 
# using convolution and max-pooling layers. Use the extracted features for 
# alphabet classification and display the test accuracy. 
```

```python
# Import required libraries

import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
# Load training and testing datasets

train_data = pd.read_csv('sign_mnist_train.csv')
test_data = pd.read_csv('sign_mnist_test.csv')

print("Training Data:", train_data.shape)
print("Testing Data:", test_data.shape)
```

```python
# Separate images and labels

X_train = train_data.drop('label', axis=1).values
y_train = train_data['label'].values

X_test = test_data.drop('label', axis=1).values
y_test = test_data['label'].values
```

```python
# Convert labels into 24 classes

classes = np.sort(np.unique(y_train))
label_map = {label: i for i, label in enumerate(classes)}

y_train = np.array([label_map[x] for x in y_train])
y_test = np.array([label_map[x] for x in y_test])

# Normalize pixel values between 0 and 1

X_train = X_train / 255.0
X_test = X_test / 255.0

# Reshape images into 28x28x1 format

X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
```

```python
# Build CNN model for feature extraction and classification

model = tf.keras.Sequential([

    # First convolution layer extracts image features
    tf.keras.layers.Conv2D(32, (3,3), activation='relu'),

    # First max pooling layer reduces dimensions
    tf.keras.layers.MaxPooling2D((2,2)),

    # Second convolution layer extracts more features
    tf.keras.layers.Conv2D(64, (3,3), activation='relu'),

    # Second max pooling layer
    tf.keras.layers.MaxPooling2D((2,2)),

    # Convert extracted features into 1D
    tf.keras.layers.Flatten(),

    # Fully connected layer
    tf.keras.layers.Dense(128, activation='relu'),

    # Output layer for 24 alphabet classes
    tf.keras.layers.Dense(24, activation='softmax')
])

# Display model summary
model.build((None, 28, 28, 1))
model.summary()
```

```python
# Compile the CNN model

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# Train the CNN model

model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128
)
```

```python
# Evaluate the model using testing data

test_loss, test_accuracy = model.evaluate(X_test, y_test)

# Display test accuracy

print("Test Accuracy: {:.2f}%".format(test_accuracy * 100))
```
