```python
# 12. Develop a Feedforward Neural Network using TensorFlow/Keras to 
# classify handwritten digits. Normalize the pixel values, flatten the images, 
# train the model and display testing accuracy. Use -- Handwritten 
# MNIST 
```

```python
# Import required libraries
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
# Load MNIST dataset directly from Keras
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Display dataset shapes
print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
# Normalize pixel values between 0 and 1
X_train = X_train / 255.0
X_test = X_test / 255.0

# Flatten 28x28 images into 784 values
X_train = X_train.reshape(-1, 784)
X_test = X_test.reshape(-1, 784)

# Display new shapes
print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
```

```python
# Create Feedforward Neural Network
model = tf.keras.Sequential([
    # Input layer
    tf.keras.layers.Input(shape=(784,)),

    # Hidden layers
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(64, activation='relu'),

    # Output layer for 10 digits (0-9)
    tf.keras.layers.Dense(10, activation='softmax')
])

# Display model summary
model.summary()
```

```python
# Compile the FNN model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
# Train the model
model.fit(X_train, y_train, epochs=10, batch_size=128)
```

```python
# Evaluate the model on testing data
test_loss, test_accuracy = model.evaluate(X_test, y_test)

# Display testing accuracy
print("Testing Accuracy:", test_accuracy * 100, "%")
```
